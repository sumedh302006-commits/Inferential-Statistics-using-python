<h1> Q.2) Check the goodness of fit for the following data mapping number of hours a student studies v/s the expected number of hours at 5% los
<h2> Observed Data (in hours) : 8, 6, 10, 7, 8, 11, 9
<h1> Expected data (in hours) : 9, 8, 11, 8, 10, 7, 6

```python
import numpy as np
from scipy.stats import chisquare

observed = np.array([8, 6, 10, 7, 8, 11, 9])
expected = np.array([9, 8, 11, 8, 10, 7, 6])

# Chi-Square Test
chi_stat, p_value = chisquare(f_obs=observed, f_exp=expected)

print("Chi-Square Statistic:", chi_stat)
print("p-value:", p_value)

# at 5% level of significance
alpha = 0.05

if p_value < alpha:
    print("Reject H0: Observed data does not fit expected distribution.")
else:
    print("Fail to Reject H0: Observed data fits expected distribution.")
```

<h1> Q.2) What is the probability of 2 heads in 6 coin toses ? 
Fit the binomial distribution for the coin tosses and plot it

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom

n = 6   # number of trials
p = 0.5 # probability of head
k = 2   # number of heads

#Prob.  of exactly 2 heads
prob_2_heads = binom.pmf(k, n, p)
print("Probability of 2 heads in 6 tosses:", prob_2_heads)

x = np.arange(0, n+1)  # possible outcomes (0 to 6 heads)
pmf_values = binom.pmf(x, n, p)

plt.bar(x, pmf_values, color='lightgreen', edgecolor='black')
plt.title("Binomial Distribution: 6 Tosses, p=0.5")
plt.xlabel("Number of Heads")
plt.ylabel("Probability")
plt.xticks(x)
plt.show()
```
